# Libraries & Paths

```python
import random, os, json, warnings
import numpy as np
import scanpy as sc
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.sparse as sp
from pathlib import Path
import decoupler as dc

# Seed of use for reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)


# Setting the path of input & output data 
data_dir   = Path('../data')   
output_dir = Path('../outputs')
output_dir.mkdir(parents=True, exist_ok=True)
figures_dir = Path("../figures")
figures_dir.mkdir(parents=True, exist_ok=True)
filtered_anndata = output_dir / "filtered_anndata.h5ad"


warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid", palette="muted", font_scale=1.1)
print ("Loaded")
```

# Loading AnnData

```python
# Loading Annadata
adata = sc.read_h5ad(filtered_anndata)


# Reloading the anndata for re-checking its values and shape while also printing the expression value range of the samples in the expression matrix.
print(f"AnnData shape: {adata.shape}")
print(f"obs columns: {list(adata.obs.columns)}")
print(f"First 5 gene names: {adata.var_names[:5].tolist()}")
X_sample = adata.X[:100, :100]
if sp.issparse(X_sample):
    X_sample = X_sample.toarray()
print(f"Expression value range of samples: [{X_sample.min():.2f}, {X_sample.max():.2f}]")
```

# Per-cell Pathway Activity Score

```python
# Loading PROGENy pathway-gene weight matrix where top = 500 (default) responsive genes per pathway are used.
progeny= dc.op.progeny(organism='human', top=500) # [9]
print(f"PROGENy dataset shape: {progeny.shape}")
print(f"Column info: {progeny.columns.tolist()}")
print(f"Total pathways: {progeny['source'].nunique()}")
print(f"Naming of Pathways: {sorted(progeny['source'].unique())}")

display(progeny.head(10))
```

**Comment:** PROGENy has `14 fixed pathways` (Androgen, EGFR, Estrogen, Hypoxia, JAK-STAT, MAPK, NFkB, PI3K, TGFb, TNFa, Trail, VEGF, WNT and p53), while the retrieved network has a shape of `6,463 rows and 4 columns` (source = pathway name, target = target gene, weight, and adjusted p-value (padj)). 

The `top=500` parameter only controls how many of the top-ranked target genes per pathway are retained in each pathway footprint and does not change the total number of available pathways and was explicitly selected when the PROGENy network was retrieved through decoupler [9].

## Overlap check PROGENy \ AnnData

```python
progeny_genes = set(progeny["target"].unique())
adata_genes = set(adata.var_names.astype(str))
overlap_genes = progeny_genes & adata_genes

print(f"PROGENy: {len(progeny_genes):,}")
print(f"AnnData has {len(adata_genes):,} genes")
print(f"Overlap: {len(overlap_genes):,} genes ({len(overlap_genes)/len(progeny_genes)*100:.1f}% of PROGENy genes found)")
```

**Comment:** Of the 10,019 genes remaining after filtering, 3,120 overlap with the PROGENy gene set, representing 59.1% of the unique PROGENy genes identified. 

Those pathway scores will be added as additional features, possibly providing biological information about the activity of major signaling pathways that may not be directly captured from individual gene-expression values alone.

```python
# Checking the overlap separately for each pathway, since the global percentage
# does not show whether some pathway has substantially lower target-gene coverage.
coverage_per_pathway = (
    progeny.assign(present=progeny["target"].isin(adata_genes))
    .groupby("source", observed=True)
    .agg(
        total_targets=("target", "nunique"),
        overlapping_targets=("present", "sum")
    )
)
coverage_per_pathway["coverage_fraction"] = (
    coverage_per_pathway["overlapping_targets"] / coverage_per_pathway["total_targets"]
)
coverage_per_pathway["coverage_percent"] = 100 * coverage_per_pathway["coverage_fraction"]

display(coverage_per_pathway.sort_values("coverage_percent"))
```

```python
# Running decoupler MLM to compute per-cell PROGENy pathway activity scores
# tmin=5 default (minimum number of targers per source--> fewer targets removed) [10]
dc.mt.mlm(data=adata,net=progeny, tmin=5,raw=False,verbose=True) 

# Extracting pathway scores from adata.obsm
# Cells --> rows
# Pathways --> columns
pathway_scores = dc.pp.get_obsm(adata,key="score_mlm") #[11]

# Printing info regarding the final shape of the pathway score matrix of all  genes
# while displaying the form of the matrix - first 5 rows.
print(f"\nPathway score matrix shape: {pathway_scores.shape}")
display(pathway_scores.to_df().head())
```

**Comment:** `MLM` produced a **(16,291 x 14)** score matrix, meaning one activity score for each of the 14 PROGENy pathways in every cell. Positive and negative MLM scores represent relative pathway activity according to the direction and weights of the signed PROGENy regulatory footprint, rather than simply high or low expression of the pathway genes themselves.

## QC for Pathway Scores

```python
# Checking for missing values
pathway_qc = pathway_scores.to_df()
print("Missing values per pathway:")
print(pathway_qc.isna().sum())

# Checking pathway score variance
variances = pathway_qc.var().sort_values()
print("\nPathway variances:")
print(variances.round(4)) 
# All of them have reasonable variances larger than > 0.01
```

**Comment:** There were `no missing values` identified in any of the 14 pathway scores, but their `variance`s seemed to be `not uniform`. Specifically, `JAK-STAT` showed a clearly larger variance compared with the remaining pathways (`10.35 in comparison to a range of approximately 0.73 to 2.22`). However, this larger variance does not by itself indicate a data-quality problem, but it represents a clear scale difference that should be considered when the pathway scores are later combined with other feature groups. 

Therefore, the pathway was retained, since its larger variance alone was not considered a sufficient reason for its removal, while this difference should be taken into account in later steps and especially during the modeling task.

## Set of Visualizations for the Pathway Scores

```python
# Creating Violin plots for each PROGENy pathway activity for each response group.
pathway_cols = pathway_qc.columns.tolist()
pathway_qc["response_label"] = adata.obs["response_label"].values

# Plotting (subplot layouts, response colors, one violin plot per pathway, zero lining --> visual choice)
n_pathways = len(pathway_cols)
ncols = 4
nrows = (n_pathways + ncols - 1) // ncols
fig, axes = plt.subplots(nrows, ncols, figsize=(ncols * 4, nrows * 3.2))
axes = axes.flatten()
colors = {"Responder": "#2196F3","Non-responder": "#FF5722"}
for i, pathway in enumerate(pathway_cols):
    ax = axes[i]
    sns.violinplot( data=pathway_qc,x="response_label",y=pathway,palette=colors,ax=ax,inner="box",linewidth=0.8)
    ax.set_title(pathway, fontweight="bold", fontsize=11)
    ax.set_xlabel("")
    ax.set_ylabel("Activity score")
    ax.axhline(0, color="gray", linestyle="--", linewidth=0.8, alpha=0.7)
    ax.tick_params(axis="x", labelsize=8)


# Hiding empty subplot spaces 
for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)
    
plt.suptitle("PROGENy Pathway Activity by Response Group",fontsize=13,fontweight="bold",y=1.01)
plt.tight_layout()
plt.savefig(figures_dir / "violin_pathway_response.png", bbox_inches="tight")
plt.show()
```

**Comment:** Creation of a violin plot for visualization purposes and practically for inspection at the single-cell level, since thousands of cells originate from the same biological samples and are therefore not independent observations.

- Not used for selecting pathways for the predictive model, and all 14 predefined PROGENy pathways are retained for the downstream analysis.

**Observations:** `JAK-STAT and EGFR show the clearest visual differences` between the response groups, while smaller shifts can also be visually observed for pathways such as PI3K, TNFa and MAPK. However, the distributions of many pathways remain highly overlapping. Also, the wider distribution observed for JAK-STAT is consistent with its considerably larger variance identified during the previous QC inspection.


```python
# Creating a heatmap of mean PROGENy pathway activity
# for each response and timepoint combination.
# Firstly, pathway scores are averaged per biological sample so that samples
# with larger numbers of cells do not contribute more weight to the group mean.
pathway_qc["timepoint"] = adata.obs["timepoint"].values
pathway_qc["patient_timepoint"] = adata.obs["patient_timepoint"].values

sample_pathways = ( pathway_qc.groupby(["patient_timepoint", "response_label", "timepoint"], observed=True)[pathway_cols].mean().reset_index())

# Calculating the mean of the biological-sample means for each response + timepoint group
mean_by_group = (sample_pathways.groupby(["response_label", "timepoint"], observed=True)[pathway_cols].mean())
mean_by_group.index = [f"{response} | {timepoint}" for response, timepoint in mean_by_group.index]

# Plotting
fig, ax = plt.subplots(figsize=(14, 4))
sns.heatmap(mean_by_group, cmap="RdBu_r", center=0, annot=True, fmt=".2f",
            linewidths=0.4, ax=ax, cbar_kws={"label": "Mean sample-level pathway activity"})
ax.set_title("Mean Sample-Level PROGENy Pathway Activity - Response x Timepoint")
ax.set_xlabel("Pathway")
plt.tight_layout()
plt.savefig(figures_dir / "pathway_heatmap_response_timepoint.png", bbox_inches="tight")
plt.show()
```

**Observations:** The heatmap was also created only for inspection purposes, and it showed in general that for several pathways the visual differences between the response groups appear to remain in the same direction across Pre and Post samples.

*More specifically, though*:

- JAK-STAT presents the clearest descriptive difference between the response groups, with higher mean activity in Non-responders compared with Responders at both Pre (4.39 vs 2.44) and Post (3.49 vs 1.82) timepoints.

- EGFR and PI3K also show higher mean activity in Non-responders at both timepoints, although with smaller differences, while MAPK presents lower mean activity in Non-responders compared with Responders at both timepoints.

- For the remaining pathways, the differences between the response groups are generally smaller or less consistent across the two timepoints.

```python
# Since in EDA we have different UMAPs for each feauture group
# Creation of UMAP coloured by top-variance pathways to:
# visualize whether the pathways with the highest cell-level variance
# follow the cellular structure already visible in the gene-expression UMAPs.

# Selecting top five pathways with the highest variance
top5_pathways = (variances.sort_values(ascending=False).head(5).index.tolist())

# Check that pathway_df and adata contain the same cells
missing_cells = adata.obs_names.difference(pathway_qc.index)
print (missing_cells)

# Adding pathway scores temporarily to adata.obs.
# Reindexing aligns the pathway scores with the exact cell order in adata.
# Creating the PCA and the UMAP for each pathway
temporary_columns = []

for pathway in top5_pathways:
    column_name = f"PROGENy_{pathway}"
    adata.obs[column_name] = (pathway_qc[pathway].reindex(adata.obs_names).astype(float))
    temporary_columns.append(column_name)

sc.tl.pca(adata,n_comps=50,random_state=SEED)
sc.pp.neighbors(adata,n_neighbors=15,n_pcs=20,random_state=SEED)
sc.tl.umap(adata,random_state=SEED)

# Plotting response status and the top five pathway activity scores
color_keys = ["response_label"] + temporary_columns
sc.pl.umap(adata,color=color_keys,ncols=3,wspace=0.35,cmap="viridis",show=False)
plt.suptitle("UMAP Coloured by Response and Top-Variance PROGENy Pathways",fontsize=20,y=1.02)
plt.savefig(figures_dir / "umap_pathways.png",dpi=150,bbox_inches="tight")
plt.show()
plt.close()


# Removing only the temporary pathway columns used and added for this plot
adata.obs.drop(columns=temporary_columns,inplace=True,errors="ignore")
```

**Comment:** Across the UMAP, the `response visualization` shows a substantial `overlap between Responder and Non-responder cells`, without a clear global separation between the two response groups, with the visualization being generated using the previously (Task A) defined dimensionality-reduction parameters, without further parameter adjustment based on the response labels. 

Regarding the pathway scores, they show different distributions across specific areas of the embedding, but they do not appear to follow a simple separation according to response, possibly suggesting that the observed pathway variation may also be related to other conditions, such as underlying cellular states or cellular composition, rather than response alone.

```python
# Creating and Calculating pathway score pearson correlation matrix
# in order to examine whether some PROGENy pathway scores contain similar information across cells.
# Selecting only the pathway activity score columns.
pathway_corr_df = pathway_qc[pathway_cols].copy()

# Calculating pairwise Pearson correlations between pathway scores
corr = pathway_corr_df.corr(method="pearson")

# Masking the upper triangle to avoid displaying each correlation twice since the matrix is symmetrical. 
mask = np.triu(np.ones_like(corr, dtype=bool),k=1)

# Plotting 
fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(corr,mask=mask,cmap="RdBu_r",center=0,vmin=-1,vmax=1,annot=True,fmt=".2f",linewidths=0.4,square=True, \
    ax=ax,cbar_kws={"label": "Pearson correlation"})
ax.set_title("PROGENy Pathway Score Pearson Correlation Matrix",fontsize=12)
plt.tight_layout()
plt.savefig(figures_dir / "pathway_pearson_correlation_matrix.png",dpi=150,bbox_inches="tight")
plt.show()
plt.close()
```

**Comment:** The Pearson correlation matrix is used as an exploratory check of how similarly the 14 pathway scores vary across cells.

**Observations:** The strongest correlation observed across the pathway scores is the negative correlation between NFkB and TNFa (r = -0.58), followed by the negative correlation between EGFR and MAPK (r = -0.42). Some smaller positive correlations can also be observed, such as NFkB with TGFb (r = 0.32) and Hypoxia (r = 0.31), while most of the remaining pathway pairs show relatively weak correlations closer to zero. 

**Conclusion:** Although some pathways appear to share partially related activity patterns across cells, the correlation matrix does not indicate a generally strong redundancy across the complete 14-pathway feature block. 

> `Worth mentioning`, despite presenting the largest variance in the previous QC analysis, JAK-STAT does not show a strong correlation with any of the remaining pathway scores, suggesting that its larger variability is not simply explained by a strong linear relationship with another pathway.



## New AnnData creation Genes + Pathways

```python
# Making a new naming (prefix addition PROGENy_) of the pathways in order to search for them and add them
# while also aligning pathway scores with the exact cell order in Anndata
pathway_df = pathway_qc.drop(columns=["response_label", "timepoint", "group"],errors="ignore").copy()
pathway_df.columns = [col if col.startswith("PROGENy_") else f"PROGENy_{col}"for col in pathway_df.columns]
pathway_df = pathway_df.reindex(adata.obs_names)

# Checking if missing values where introduced during alignment 
if pathway_df.isna().any().any():
    raise ValueError("Missing values found.")

# Creating a new anndata and adding the pathway activity scores
adata_pathways = adata.copy()
adata_pathways.obs = adata_pathways.obs.join(pathway_df)

print("New columns added:")
for col in adata_pathways.obs.columns:
    if col.startswith("PROGENy_"):
        print(col)

print(f"\nNew AnnData shape: {adata_pathways.shape}")
print("Number of pathway columns:",sum( col.startswith("PROGENy_")for col in adata_pathways.obs.columns))
```

```python
# Saving new Anndata
pathway_anndata = output_dir /"filtered_anndata_pathways.h5ad"
adata_pathways.write_h5ad(pathway_anndata)
print(f"\nSaved extended AnnData: {pathway_anndata}")

# Reloading
print("----------------------------------------------------")
adata = sc.read_h5ad(pathway_anndata)
# Reloading the anndata for re-checking its values and shape.
print(f"AnnData shape: {adata.shape}")
print(f"obs columns: {list(adata.obs.columns)}")
```
